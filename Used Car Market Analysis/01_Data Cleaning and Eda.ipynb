# Fix clear data entry errors

Rather than deleting these rows outright (the rest of each row is still
useful — body type, features, specs, etc.), the *bad field only* is set to
`NaN` so it gets repaired by the same imputation logic used for genuinely
missing values.

- **Odometer (`km`) > 2,000,000** — physically impossible for a used car.
- **`listed_price`** flagged as an error only when it is *both* an extreme
  outlier vs. other listings of the same oem+model (>8x the group median)
  *and* above ₹1 crore *and* the brand isn't a known luxury/exotic make
  (so genuine Ferrari/Lamborghini/Mercedes-S-Class prices are left alone).


```python
km_bad = df['km'] > 2_000_000
print(f"km values > 2,000,000 (odometer errors): {km_bad.sum()} rows")
display(df.loc[km_bad, ['oem', 'model', 'myear', 'km', 'listed_price']])
df.loc[km_bad, 'km'] = np.nan
```

```python
luxury_oems = {'ferrari', 'lamborghini', 'bentley', 'aston martin', 'rolls-royce',
               'maserati', 'porsche', 'mclaren', 'mercedes-benz', 'bmw', 'audi',
               'jaguar', 'land rover', 'lexus', 'volvo'}

med_price = df.groupby(['oem', 'model'])['listed_price'].transform('median')
price_bad = (df['listed_price'] > 2 * med_price) & (df['listed_price'] > 1e7) & (~df['oem'].isin(luxury_oems))

print(f"listed_price data-entry errors: {price_bad.sum()} rows")
display(df.loc[price_bad, ['oem', 'model', 'myear', 'km', 'listed_price']])
df.loc[price_bad, 'listed_price'] = np.nan
```

# Merging duplicate color /exterior color columns 
 cross fill each from other before imputing 

```python
before_color_na = df['Color'].isna().sum()
before_ext_na = df['exterior_color'].isna().sum()

same = (df['Color'] == df['exterior_color']).sum()
print(f"Color == exterior_color in {same} of {df['Color'].notna().sum()} rows where both present")

df['Color'] = df['Color'].fillna(df['exterior_color'])
df['exterior_color'] = df['exterior_color'].fillna(df['Color'])

print(f"Color missing: {before_color_na} -> {df['Color'].isna().sum()}")
print(f"exterior_color missing: {before_ext_na} -> {df['exterior_color'].isna().sum()}")
```

# imputation with similar values -- rows , 
oem + model + varian using or whatever best available 

```python
GROUP_LEVELS = [
    ['oem', 'model', 'variant'],
    ['oem', 'model', 'fuel', 'transmission'],
    ['oem', 'model'],
    ['oem', 'body'],
]

def hierarchical_fill_numeric(frame, col):
    result = frame[col].copy()
    n_before = result.isna().sum()
    level_counts = []
    for keys in GROUP_LEVELS:
        if result.isna().sum() == 0:
            break
        fill_vals = frame.groupby(keys)[col].transform('median')
        newly = result.isna() & fill_vals.notna()
        level_counts.append(newly.sum())
        result = result.fillna(fill_vals)
    remaining = result.isna().sum()
    if remaining:
        result = result.fillna(frame[col].median())
    return result, n_before, level_counts, remaining


def hierarchical_fill_categorical(frame, col):
    result = frame[col].copy()
    n_before = result.isna().sum()
    level_counts = []

    def mode_or_nan(s):
        m = s.mode()
        return m.iat[0] if not m.empty else np.nan

    for keys in GROUP_LEVELS:
        if result.isna().sum() == 0:
            break
        mode_map = frame.groupby(keys)[col].agg(mode_or_nan)
        if len(keys) > 1:
            fill_vals = frame.set_index(keys).index.map(mode_map)
        else:
            fill_vals = frame[keys[0]].map(mode_map)
        fill_vals = pd.Series(fill_vals, index=frame.index)
        newly = result.isna() & fill_vals.notna()
        level_counts.append(newly.sum())
        result = result.fillna(fill_vals)
    remaining = result.isna().sum()
    if remaining:
        global_mode = frame[col].mode()
        if not global_mode.empty:
            result = result.fillna(global_mode.iat[0])
    return result, n_before, level_counts, remaining
```

```python
numeric_cols = [
    'km', 'listed_price', 'No of Cylinder', 'Valves per Cylinder', 'Length', 'Width',
    'Height', 'Wheel Base', 'Front Tread', 'Rear Tread', 'Kerb Weight', 'Seats',
    'Turning Radius', 'Top Speed', 'Acceleration', 'Doors', 'Cargo Volume',
    'Alloy Wheel Size', 'Max Power Delivered', 'Max Power At', 'Max Torque Delivered',
    'Max Torque At',
]

categorical_cols = [
    'body', 'Valve Configuration', 'Gear Box', 'Drive Type', 'Steering Type',
    'Front Brake Type', 'Rear Brake Type', 'Tyre Type', 'Engine Type',
    'Fuel Suppy System', 'Color', 'exterior_color',
]
```

```python
numeric_summary = []
for col in numeric_cols:
    filled, n_before, level_counts, remaining_before_global = hierarchical_fill_numeric(df, col)
    df[col] = filled
    lc = level_counts + [0] * (4 - len(level_counts))
    numeric_summary.append([col, n_before, *lc, remaining_before_global])

# round count-like fields back to whole numbers after median imputation
int_like_cols = ['No of Cylinder', 'Valves per Cylinder', 'Seats', 'Doors', 'Alloy Wheel Size']
for col in int_like_cols:
    df[col] = df[col].round(0)

numeric_report = pd.DataFrame(
    numeric_summary,
    columns=['column', 'missing_before', 'L1_model+variant', 'L2_model+fuel+trans', 'L3_model', 'L4_oem+body', 'global']
)
numeric_report
```

```python
categorical_summary = []
for col in categorical_cols:
    filled, n_before, level_counts, remaining_before_global = hierarchical_fill_categorical(df, col)
    df[col] = filled
    lc = level_counts + [0] * (4 - len(level_counts))
    categorical_summary.append([col, n_before, *lc, remaining_before_global])

categorical_report = pd.DataFrame(
    categorical_summary,
    columns=['column', 'missing_before', 'L1_model+variant', 'L2_model+fuel+trans', 'L3_model', 'L4_oem+body', 'global']
)
categorical_report
```
